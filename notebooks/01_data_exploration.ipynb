# SyncNet Experiment 01 — Exploratory Data Analysis (EDA)

**Objective**: Analyze class balance (bot vs human), profile metadata distributions, activity frequencies, and interaction graph topology.
**Input**: `data/raw/users.jsonl`, `data/raw/posts.jsonl`, `data/raw/edges.jsonl`

```python
import os
import json
import pandas as pd
import numpy as np

DATA_RAW_DIR = os.path.join("..", "data", "raw")

users_file = os.path.join(DATA_RAW_DIR, "users.jsonl")
posts_file = os.path.join(DATA_RAW_DIR, "posts.jsonl")
edges_file = os.path.join(DATA_RAW_DIR, "edges.jsonl")

users = [json.loads(line) for line in open(users_file, "r", encoding="utf-8")]
posts = [json.loads(line) for line in open(posts_file, "r", encoding="utf-8")]
edges = [json.loads(line) for line in open(edges_file, "r", encoding="utf-8")]

df_users = pd.DataFrame(users)
df_posts = pd.DataFrame(posts)
df_edges = pd.DataFrame(edges)

print(f"Total Accounts : {len(df_users)}")
print(f"Total Posts    : {len(df_posts)}")
print(f"Total Edges    : {len(df_edges)}")
```

## 1. Ground Truth Label Distribution

```python
label_counts = df_users["label"].value_counts()
print("Class Distribution (0: Human, 1: Bot):")
print(label_counts)
print(f"Bot Ratio: {df_users['label'].mean():.2%}")
```

## 2. Profile Attributes Analysis

```python
print("--- Followers Count Summary by Class ---")
print(df_users.groupby("label")["followers_count"].describe())

print("\n--- Following Count Summary by Class ---")
print(df_users.groupby("label")["following_count"].describe())
```

## 3. Interaction Topology Analysis

```python
print("Relation Type Distribution:")
print(df_edges["relation"].value_counts())

# Check graph connectivity
unique_nodes = set(df_edges["source"]).union(set(df_edges["target"]))
print(f"Connected Unique Nodes in Graph: {len(unique_nodes)} / {len(df_users)}")
```
