# SyncNet Experiment 04 — Behavioral Feature Engineering & Matrix Inspection

**Objective**: Inspect the 12 numerical behavioral features extracted from account metadata and post engagement metrics.
**Input**: `data/features/behavioral_features.parquet`

```python
import os
import pandas as pd
import numpy as np

DATA_FEAT_DIR = os.path.join("..", "data", "features")
df_feat = pd.read_parquet(os.path.join(DATA_FEAT_DIR, "behavioral_features.parquet"))

print(f"Behavioral Feature Matrix Dimensions: {df_feat.shape}")
df_feat.head(5)
```

## Feature Summary Statistics by Class (Bot vs Human)

```python
numeric_cols = [c for c in df_feat.columns if c not in ["user_id", "split", "label"]]
print("Feature Means by Label:")
display_df = df_feat.groupby("label")[numeric_cols].mean().T
display_df.columns = ["Human (0)", "Bot (1)"]
print(display_df)
```
